```python
def analisar_telemetria(dados):
    print("--- INICIANDO VERIFICAÇÃO DE PRÉ-DECOLAGEM ---\n")
    
    # Extração de variáveis
    temp_int = dados.get('temperatura_interna')
    integridade = dados.get('integridade_estrutural')
    energia = dados.get('nivel_energia')
    pressao = dados.get('pressao_tanques')
    modulos = dados.get('modulos_criticos')
    
    # Exibindo os dados lidos
    print(f"Temperatura Interna: {temp_int}°C")
    print(f"Integridade Estrutural: {'OK (1)' if integridade == 1 else 'FALHA (0)'}")
    print(f"Nível de Energia: {energia}%")
    print(f"Pressão dos Tanques: {pressao} psi")
    print(f"Módulos Críticos: {'Operantes' if modulos else 'Inoperantes'}\n")

    # Regras de Negócio / Verificações
    if integridade != 1:
        return "STATUS: DECOLAGEM ABORTADA (Motivo: Integridade estrutural comprometida)"
    if not modulos:
        return "STATUS: DECOLAGEM ABORTADA (Motivo: Módulos críticos inoperantes)"
    if energia < 85.0:
        return "STATUS: DECOLAGEM ABORTADA (Motivo: Energia abaixo de 85%)"
    if not (3000 <= pressao <= 4500):
        return f"STATUS: DECOLAGEM ABORTADA (Motivo: Pressão {pressao} psi fora da faixa segura)"
    if not (18 <= temp_int <= 26):
        return f"STATUS: DECOLAGEM ABORTADA (Motivo: Temperatura interna {temp_int}°C fora do padrão)"

    return "STATUS: PRONTO PARA DECOLAR"

# --- Simulação ---
telemetria_atual = {
    'temperatura_interna': 22,
    'temperatura_externa': -45,
    'integridade_estrutural': 1,
    'nivel_energia': 80,
    'pressao_tanques': 3800,
    'modulos_criticos': True
}

resultado = analisar_telemetria(telemetria_atual)
print(resultado)
```
